```python
import re
from itertools import product

class Bit:
    """A single Boolean value (0 or 1) with logical operators."""
    def __init__(self, value):
        self.value = int(bool(value))
    def __and__(self, other): return Bit(self.value & other.value)
    def __or__(self, other):  return Bit(self.value | other.value)
    def __xor__(self, other): return Bit(self.value ^ other.value)
    def __invert__(self):     return Bit(1 - self.value)
    def __eq__(self, other):  return Bit(self.value == other.value)
    def __hash__(self):       return hash(self.value)
    def __str__(self):        return str(self.value)

OPERATORS = {"AND": "&", "XOR": "^", "OR": "|", "NOT": "~", "IS": "=="}

def truthTable(expression, inputs=2):
    if inputs not in (2, 3, 4):
        raise ValueError("inputs must be 2, 3 or 4")
    names = "PQRS"[:inputs]
    python_expr = re.sub(r"\b(AND|XOR|OR|NOT|IS)\b",
                         lambda m: OPERATORS[m.group(1)],
                         expression.upper())
    # Only the letters P Q R S, operators and parentheses are allowed (no arbitrary code)
    if not re.fullmatch(r"[PQRS\s&|^~()=]+", python_expr):
        raise ValueError(f"Unsupported characters in expression: {expression!r}")

    rows = []
    for values in product([0, 1], repeat=inputs):
        env = {name: Bit(v) for name, v in zip(names, values)}
        try:
            result = eval(python_expr, {"__builtins__": {}}, env)
        except (SyntaxError, NameError, TypeError, AttributeError) as error:
            raise ValueError(f"Could not evaluate {expression!r}: {error}") from None
        if not isinstance(result, Bit):
            raise ValueError(f"Could not evaluate {expression!r}")
        rows.append((values, result))

    print(f"Boolean Expression:\n  X = {expression.strip().upper()}\n")
    print("| " + " | ".join(names) + " | X |")
    print("|" + "---|" * (inputs + 1))
    for values, result in rows:
        print("| " + " | ".join(str(v) for v in values) + f" | {result} |")
```

```python
#expression for input as the boolean expression
expression = "P AND NOT (Q XOR R)"
#implementing the previously written function
truthTable(expression,3)
```
