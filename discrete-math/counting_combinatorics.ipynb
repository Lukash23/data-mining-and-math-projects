```python
import math

def permutations(n, r):
    """Number of ordered arrangements of r items chosen from n: n! / (n - r)!"""
    if not (0 <= r <= n):
        raise ValueError("Need 0 <= r <= n")
    return math.factorial(n) // math.factorial(n - r)

n, r = 5, 3
print(f"Ordered sequences of {r} from {n}: {permutations(n, r)}")
print("Check:", permutations(n, r) == math.perm(n, r))
```

```python
def combinations(n, r):
    """Number of unordered selections of r items from n: n! / (r! (n - r)!)"""
    if not (0 <= r <= n):
        raise ValueError("Need 0 <= r <= n")
    return math.factorial(n) // (math.factorial(r) * math.factorial(n - r))

print(f"Unordered selections of {r} from {n}: {combinations(n, r)}")
print("Check:", combinations(n, r) == math.comb(n, r))
```

```python
def binomial_coefficient(n, r, k):
    """Coefficient of x^r in the expansion of (x + k)^n: C(n, r) * k^(n - r)"""
    return combinations(n, r) * k ** (n - r)

n, r, k = 9, 4, -3
print(f"Coefficient of x^{r} in (x{k:+})^{n} is {binomial_coefficient(n, r, k)}")

import sympy as sp
x = sp.symbols("x")
expected = sp.Poly((x + k) ** n, x).coeff_monomial(x ** r)
print("Check:", binomial_coefficient(n, r, k) == expected, "| SymPy says", expected)
```
