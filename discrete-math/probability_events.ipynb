```python
import numpy as np # linear algebra
import math

# Creating a sample space of 2coin flips
sample_space = {'HH', 'TT','HT','TH'}



# E1: event outcome 2 heads {'HH'}
# E2: event outcome 2 tails {'TT'}
def is_heads_or_tails(outcome):  return outcome in {'HH', 'TT','HT','TH'}
def is_2heads_or_2tails(outcome):  return outcome in {'HH', 'TT'}
def is_neither(outcome): return not is_heads_or_tails(outcome)


# Defining additional event conditions
def is_2heads(outcome): return outcome == 'HH'
def is_2tails(outcome): return outcome == 'TT'

# Defining an event-detection function
def get_matching_event(event_condition, sample_space):
    return set([outcome for outcome in sample_space
                if event_condition(outcome)])

# Detecting events using event conditions
event_conditions = [is_heads_or_tails, is_2heads, is_2tails,is_neither]

for event_condition in event_conditions:
    print(f"Event Condition: {event_condition.__name__}")
    event = get_matching_event(event_condition, sample_space)
    print(f'Event: {event}\n')
```

```python
# Computing event probabilities
def compute_probability(event_condition, generic_sample_space):
    event = get_matching_event(event_condition, generic_sample_space)
    return len(event) / len(generic_sample_space)

for event_condition in event_conditions:
    prob = compute_probability(event_condition, sample_space)
    name = event_condition.__name__
    print(f"Probability of event arising from '{name}' is {prob}")
```

```python
# Compound events on the two-coin sample space {'HH', 'TT', 'HT', 'TH'}
def at_least_1_head(outcome): return 'H' in outcome
def at_least_1_tail(outcome): return 'T' in outcome
def head_and_tail(outcome): return at_least_1_head(outcome) and at_least_1_tail(outcome)  # intersection
def head_or_tail(outcome): return at_least_1_head(outcome) or at_least_1_tail(outcome)    # union

compound_conditions = [at_least_1_head, at_least_1_tail, head_and_tail, head_or_tail]

for condition in compound_conditions:
    event = get_matching_event(condition, sample_space)
    prob = compute_probability(condition, sample_space)
    print(f"{condition.__name__}: event = {sorted(event)}, P = {prob}, complement P = {1 - prob}")
```

```python
# Computing events probabilities
# Check the addition rule: P(A or B) = P(A) + P(B) - P(A and B)
p_a = compute_probability(at_least_1_head, sample_space)
p_b = compute_probability(at_least_1_tail, sample_space)
p_both = compute_probability(head_and_tail, sample_space)
p_either = compute_probability(head_or_tail, sample_space)

print(f"P(head) + P(tail) - P(head and tail) = {p_a + p_b - p_both}")
print(f"P(head or tail) = {p_either}")
print("Addition rule holds:", abs((p_a + p_b - p_both) - p_either) < 1e-12)

```

```python
# Complement of each compound event: all outcomes NOT in the event
for condition in compound_conditions:
    event = get_matching_event(condition, sample_space)
    complement = sample_space - event
    print(f"{condition.__name__}: complement = {sorted(complement)}, P(complement) = {len(complement) / len(sample_space)}")
```

```python
import math
import random

def binomial_pmf(k, n, p):
    return math.comb(n, k) * p**k * (1 - p)**(n - k)

n_flips = 10
p_head = 0.5   # fair coin

total = 0
for k in range(n_flips + 1):
    prob = binomial_pmf(k, n_flips, p_head)
    total += prob
    print(f"P({k} heads in {n_flips} flips) = {prob:.4f}")
print(f"Total probability = {total:.4f}")

exact = binomial_pmf(2, n_flips, p_head)
print(f"\nExact probability of exactly 2 heads in 10 fair flips = {exact:.4f}")

random.seed(42)
trials = 100000
hits = sum(1 for _ in range(trials)
           if sum(random.random() < p_head for _ in range(n_flips)) == 2)
print(f"Simulated ({trials} trials) = {hits / trials:.4f}")
```

2.1 Findings: Using a sample space of two coin flips, I computed the probabilities of simple and compound events and their complements. "At least one head" and "at least one tail" each have probability 0.75. Their intersection (one head and one tail) has probability 0.5, and their union covers the whole sample space (probability 1.0).

2.3 Simulation: The exact binomial probability of getting exactly 2 heads in 10 fair coin flips is about 0.0439. A simulation of 100,000 trials gave a very similar value. They differ slightly because the simulation uses a finite number of random trials.
